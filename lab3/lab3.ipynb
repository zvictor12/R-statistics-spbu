Принимаем нулевую гипотезу: дисперсии весов мальчиков и девочек равны

```r
iris <- read.table("/content/R-statistics-spbu/lab2/datasets/iris.txt", sep=",", col.names = c("Sepal_length", "Sepal_width", "Petal_length", "Petal_width", "Class"))
```

```r
run_pairwise_test <- function(test) {
col_names <- c("Sepal_length", "Sepal_width", "Petal_length", "Petal_width")

for(i in col_names)
{
  class_names <- c("Iris-setosa", "Iris-versicolor", "Iris-virginica")
  result <- matrix(NA, nrow = 2, ncol = 2)
  rownames(result) <- class_names[-1]
  colnames(result) <- class_names[-3]
  cn <- class_names
  for(j in class_names[-3])
  {
    cn <- cn[cn != j]
    x <- iris[iris$Class == j, i]
    for(k in cn)
    {
      y <- iris[iris$Class == k, i]
      p <- test(x, y)
      result[k, j] <- ifelse(p$p.value > 0.05, 1, 0)
    }
  }
  cat("===== FEATURE", i, "=====\n\n")
  print(result)
  cat("\n\n")
}}
```

```r
run_pairwise_test(ks.test)
```

Видим, что никакие классы не имеют одинаково распределенных попарно соответствующих признаков

```r
run_pairwise_test(t.test)
```

```r
run_pairwise_test(var.test)
```

Вывод показывает, что можно принять нулевую гипотезу о равенстве дисперсий всех классов у Sepal_width, а также у versicolor и virginica для petal и sepal length. Гипотеза о равенстве мат. ож. отклоняется везде

```r
library(readxl)
file <- "/content/R-statistics-spbu/lab3/datasets/surgery.xlsx"
header_rows <- read_excel(file, range = cell_rows(1:2), col_names = FALSE)

fill_na <- function(x) {
  for(i in seq_along(x)) {
    if(is.na(x[i])) x[i] <- x[i-1]
  }
  return(x)
}

header_rows[1, ] <- fill_na(as.vector(header_rows[1, ]))
new_colnames <- paste(header_rows[1, ], header_rows[2, ], sep = "_")
new_colnames <- gsub(" ", "_", new_colnames)

surgery <- read_excel(file, skip = 2, col_names = FALSE)
colnames(surgery) <- new_colnames
```

```r
surgery <- na.omit(surgery) #удалим строки с NA
```

```r
success <- sum((surgery$After_operation_V_left > surgery$Before_operation_V_left) & (surgery$After_operation_V_right > surgery$Before_operation_V_right))
```

```r
binom.test(success, nrow(surgery), p = 0.7)
```

```r
binom.test(success, nrow(surgery), p = 0.8)
```

Обе гипотезы принимаются с двусторонней альтернативой, приблизительная оценка вероятности 0.793
